```python
import pandas as pd

i = pd.read_csv(r"C:\\Users\\Korol\\Downloads\\spam_or_not_spam.csv")
df = i
```

```python
df["label"].value_counts()
```

```python
from sklearn.feature_extraction.text import CountVectorizer

spam_data = df
spam_data = spam_data.dropna(subset=["email"])
spam_data = spam_data[spam_data["email"].str.strip() != ""]
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(spam_data['email'])
print(X.shape)
```

```python
from sklearn.model_selection import train_test_split

y = spam_data['label']
X_train, X_test, y_train, y_test = train_test_split(
    X, y, train_size=0.75, test_size=0.25, stratify=y, random_state=42)
```

```python
from sklearn.naive_bayes import ComplementNB
from sklearn.metrics import accuracy_score, precision_score

model = ComplementNB(alpha=0.01)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
print(accuracy)
print(precision)
```
